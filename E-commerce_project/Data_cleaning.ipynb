```python
df.to_csv("cleaned_ecommerce_data.csv", index=False)
```

```python
cleaning_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str),
    "Missing_Count": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

print(cleaning_report)
```
